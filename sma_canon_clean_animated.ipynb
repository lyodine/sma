# SMA Implementation

```python
from typing import Literal as L, Callable
import numpy as np
import operator
import random
from numpy.random import default_rng
```

## Define Seed and Types

To make the algorithm easier to understand, we need to set up some stuff first. These include a global seed and types of data structures.

```python
#: This is the random seed.
GLOBAL_SEED: int = 24601
np.random.seed(GLOBAL_SEED)
random.seed(GLOBAL_SEED)

#: Number of solutions in the population.
N: int
#: Size of each solution; or dimension of the problem.
D: int

#: Type of a solution in $R^D$. Location of "one"
#: slime mould.
type Location = np.ndarray[tuple[L['D']],
                           np.dtype[np.float64]]

#: Type of the population in $R^{N\\times D}$.
#: Each $i^\mathrm{th}$ item is a `Location`.
type Population = np.ndarray[tuple[L['N'], L['D']],
                             np.dtype[np.float64]]

#: Fitness of one solution. Just a `float` in this case.
type Fitness = np.float64

#: Type of a vector of fitnesses. Contains one item
#: for each of the $N$ solutions.
type Fitnesses = np.ndarray[tuple[L['N']],
                            np.dtype[np.float64]]

MAX_ITERATION: int
```

## Define a Population Initialiser

Next, let's define an initialiser for the population.

Parameters `UB` and `LB` bound the search space, in each dimension, from above and below.

```python
def initialise_population(N: int,
                          dim: int,
                          UB: Location,
                          LB: Location)\
        -> Population:
    """Given parameters, initialise a Population from
    parameters.
    """
    
    if np.any(UB<LB):
        raise ValueError(
            "For each index i, UB[i]"
            "must not be less than LB[i].")

    Xs = default_rng(GLOBAL_SEED).random((N, dim))
    print(Xs*(UB - LB) + LB)
    return Xs*(UB - LB) + LB

UB: Location
"""Upper bounds for items in a solution.
Xs[i][j] must not exceed UB[j].
"""

LB: Location
"""Lower bounds for items in a solution.
Xs[i][j] must not fall below LB[j].
""";
```

## Define the Algorithm

Let's define the SMA algorithm to the author's specification. Also define whether the task is to maximise or minimise the objective function.

An `SMAReport` instance captures the output.

```python
from dataclasses import dataclass

@dataclass
class SMAReport:
    #: Solutions in each generation.  0th is earliest.
    solutions: list[Population]
    #: Fitnesses in each generation. 0th is earliest.
    fitnesses: list[Fitnesses]
    #: Best solution ever.
    best_solution: Location
    #: Best fitness ever.
    best_fitness: Fitness
    #: Best solutions in each generation.  0th is earliest.
    gen_best_solutions: list[Location]
    #: Best fitnesses in each generation.  0th is earliest.
    gen_best_fitnesses: list[Fitness]

def sma(objective: Callable[[Location], Fitness],
        is_maximising: bool,
        max_iteration: int,
        pop_ub: Location,
        pop_lb: Location,
        ub: Location,
        lb: Location,
        N: int,
        D: int,
        Z: float,) -> SMAReport:
    """
    Args:
        objective: The objective function.

        max_iteration: The number of iterations to use.

        pop_ub: Upper bound on the initial population.
        
        pop_lb: Lower bound on the initial population.

        ub: Upper bound on the search space.

        lb: Lower bound on the search space.

        N: Number of "members" of a slime mould.

        D: Dimension of the search space. Must be the size of
           :arg:`pop_up`, :arg:`pop_lb`, :arg:`ub`, and :arg:`lb`.

        Z: I don't know what it does, maybe a
           probability threshold? The original paper uses 0.03.
    """

    historical_solutions: list[Population] = []
    historical_fitnesses: list[Fitnesses] = []
    gen_best_solutions: list[Location] = []
    gen_best_fitnesses: list[Fitness] = []

    Xs = initialise_population(N, D, pop_ub, pop_lb)

    #: The best solution.
    best_solution: Location
    #: Fitness of the best solution.
    best_fitness: Fitness

    # Initialise :var:`best_fitness` as the worst
    #   possible fitness. If algorithm is maximising,
    #   then best_fitness starts small. Otherwise,
    #   best_fitness starts large.
    if is_maximising:
        best_fitness = np.float64(np.finfo(np.float64).min)
    else:
        best_fitness = np.float64(np.finfo(np.float64).max)

    #: N-vector. S[i] is the fitness of the i^th solution.
    S: Fitnesses

    for t in range(max_iteration):
        # Bound each Xs[i][j] to between UB[j] and LB[j]

        for i in range(len(Xs)):
            # Xs[i,:] is the ith slice
            _ub_violation_mask = Xs[i,:] > ub

            Xs[i,:] = np.multiply(Xs[i,:], ~_ub_violation_mask) +\
                          np.multiply(ub, _ub_violation_mask)

            _lb_violation_mask = Xs[i,:] < lb
            Xs[i,:] = np.multiply(Xs[i,:], ~_lb_violation_mask) +\
                          np.multiply(lb, _lb_violation_mask)

        # Calculate fitnesses. S[i] is fitness(X[i,:])
        S = np.apply_along_axis(objective, 1, Xs)

        # Algorithm 1: Calculate the W by Eq. (2.5).
        # small_dict is an ordered list. It maps
        #   each index `smell_index: int` to the
        #   smell_index^{nd} best solution.
        # smell_indices: (dim,) of int
        # smells: (dim,) of floats

        # Implemented according to the canonical implementation.
        # The best is at index 0; the worst is at index N.
        # As such, sort according to :var:`IS_MAXIMISING`:
        smell_dict = np.array(sorted(list(enumerate(S)),  # type: ignore
                                     key=lambda x: x[1],
                                     reverse=True if is_maximising\
                                                  else False))

        weights = np.zeros(shape=(N, D))
        bF: np.float64 = smell_dict[0][1]
        wF: np.float64 = smell_dict[-1][1]

        # Partition smell_dict into two roughly-equal-sized
        #   matrices containing the better / worse halves.
        smell_dict_condition = smell_dict[0:int(N/2)]
        smell_dict_others = smell_dict[int(N/2):-1]

        # This variable is named S in the canonical implementation.
        # However, in the paper S means something different.
        # This is name is to avoid ambiguity.
        # `eps` is to avoid division by zero if bF=wF.
        #   The number is so infinitesimally small that we
        #   could probably ignore the probability that
        #   bF and wF differ by exactly eps.
        S_eps = best_fitness - wF + np.finfo(np.float64).eps # was bF in paper ... maybe try to be more aggressive?

        for j in range(0, D):
            for smell_index, smell_order in smell_dict_condition:
                weights[int(smell_index), j] = 1 + random.random()*np.log10((bF-smell_order)/(S_eps)+1)

            for smell_index, smell_order in smell_dict_others:
                weights[int(smell_index), j] = 1 - random.random()*np.log10((bF-smell_order)/(S_eps)+1)

        if is_maximising:
            is_better_than = operator.gt
        else:
            is_better_than = operator.lt

        if is_better_than(bF, best_fitness):
            # Recall that the 0th solution is best.
            best_fitness = smell_dict[0][1]
            best_solution = Xs[int(smell_dict[0][0])]
            
        # Replacing t in original with t+1 here. This is because
        # Matlab indices begin as 0.
        a = np.arctanh(1-(t+1)/max_iteration)  # Code says it's (2.4)
        b = 1 - (t+1)/max_iteration

        historical_solutions.append(Xs.copy())
        historical_fitnesses.append(S.copy())
        gen_best_solutions.append(Xs[int(smell_dict[0][0])].copy())
        gen_best_fitnesses.append(smell_dict[0][1].copy())

        for i in range(N):
            rand=random.random()
            # The code ... uses rand directly?
            if rand < Z: # Code says it's (2.7)
                Xs[i,:] = (pop_ub-pop_lb)*random.random() + pop_lb
            else:
                # AllFitness in code is S here
                p = np.tanh(np.abs(S[i] - best_fitness))
                vb = np.random.uniform(low=-a, high=a, size=(D,)) # Code says it's 2.3
                vc = np.random.uniform(low=-b, high=b, size=(D,))

            for j in range(D):
                r = random.random()
                A = random.randint(0, N-1)
                B = random.randint(0, N-1)
                
                if r < p:
                    Xs[i, j] = best_solution[j] + vb[j] * (weights[i, j]*Xs[A,j]-Xs[B,j])
                else:
                    Xs[i, j] = vc[j] * Xs[i,j]


    result: SMAReport = SMAReport(
        solutions=historical_solutions,
        fitnesses=historical_fitnesses,
        best_solution=best_solution,
        best_fitness=best_fitness,
        gen_best_solutions=gen_best_solutions,
        gen_best_fitnesses=gen_best_fitnesses,
    )
    return result
```

## Define the Problem

Time for the fun stuff. Let's define some parameters then run the algorithm.

```python
import ofunc

#: The objective function
OBJECTIVE: Callable[[Location], Fitness] =\
    lambda x: ofunc.himmelblau(x[0],x[1])  #type: ignore

N = 200
D = 2
POP_UB = np.ones(shape=(D,)) * -1
POP_LB = np.ones(shape=(D,)) * -4
UB = np.ones(shape=(D,)) * 7
LB = np.ones(shape=(D,)) * -7
IS_MAXIMISING: bool = False
Z = 0.03
MAX_ITERATION = 200
```

## Run the Algorithm

With everything set up, running the algorithm is easy. Do it now.

```python
report: SMAReport = sma(objective = OBJECTIVE,
                        is_maximising = IS_MAXIMISING,
                        max_iteration = MAX_ITERATION,
                        pop_ub=POP_UB,
                        pop_lb=POP_LB,
                        ub=UB,
                        lb=LB,
                        N=N,
                        D=D,
                        Z=Z)
```

## Visualise

Let's look at the output next. We see that the algorithm is able to arrive at a decent solution, but there is more to it. Let's plot the fitness curve next.

The following cells contain code (`ofig`, `ofunc`, `oplot`) from a developmental branch of [ograph](https://pypi.org/project/ograph/).

```python
print(f"Best solution is {report.best_solution},"
      f" with fitness {report.best_fitness}")
```

### Plot Best Fitnesses

Plotting the best fitnesses across generations, we see that a good fitness was found early. It appears that the algorithm continued to explore other regions of the search space, before re-converging to a good solution. The next cell has a better visualisation of the process.

```python
import matplotlib.pyplot as plt
%matplotlib inline
%config InlineBackend.figure_format='retina'

mort = report.gen_best_fitnesses
plt.xlabel("Generation")
plt.ylabel("Fitness")

plt.scatter(range(len(mort)),
              mort,
              s=12,
              color = "tab:blue",
              facecolors="white",)


plt.vlines(x=range(len(mort)),
           ymin=np.repeat(a=report.best_fitness, repeats=len(mort)),
           ymax=mort,
           zorder=-9,
           linewidth=0.8)

plt.hlines(report.best_fitness,
           xmin=0,
           xmax=len(mort),
           color="tab:green", 
           label="Best fitness ever")

plt.plot([],
         [],
         color="tab:blue",
         label="Best fitness in generation")

plt.legend();
```

### Plot the Learning Process

With `oplot`, let's plot all "members" of the slime mould across generations. The colour of a dot indicates its fitness: green is good, red means bad.

```python
import oplot as oplot
import matplotlib as mpl

oplot.contour(lambda x, y: OBJECTIVE((x, y)), # type: ignore[reportArgumentType]
              x_range=(LB[0]-1, UB[0]+1),
              y_range=(LB[1]-1, UB[1]+1))

for i in range(len(report.solutions)):
    _xs, _ys = np.array(report.solutions[i]).T
    _intensity = i/len(report.solutions)
    plt.scatter(_xs, _ys, s=1,
                color=mpl.colormaps['RdYlGn'](_intensity))  #type: ignore
plt.axis('off')
plt.savefig("sma.svg", bbox_inches='tight')
```

### Animate the Learning Process

The following cell will animate the learning process and save it as a `.gif` file. The function has been optimised to reused the same canvas, but it can still be a bit expensive to run.

```python
from collections import deque
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation, PillowWriter
from matplotlib.collections import Collection
import numpy as np

def animate_and_save(file_name: str = "sma_animation.gif") -> None:
    fig, ax = plt.subplots()

    x_rect_lb = LB[0]
    y_rect_lb = LB[1]
    x_rect_ub = UB[0]
    y_rect_ub = UB[1]
    ax.set_xlim(x_rect_lb, x_rect_ub)
    ax.set_ylim(y_rect_lb, y_rect_ub)


    #: Keeping type correctness at all cost.
    artists_to_remove: deque[Collection] = deque()
    oplot.contour(lambda x, y: OBJECTIVE((x, y)),  # type: ignore
                  x_range=(LB[0]-1, UB[0]+1),
                  y_range=(LB[1]-1, UB[1]+1))


    def animate(i):
        while artists_to_remove:
            artists_to_remove.pop().remove()

        ax.set_xlim(x_rect_lb, x_rect_ub)
        ax.set_ylim(y_rect_lb, y_rect_ub)
        _xs, _ys = np.array(report.solutions[i]).T
        _intensity = i/len(report.solutions)
        new_artist = plt.scatter(
            _xs,
            _ys,
            s=1,
            color=mpl.colormaps['RdYlGn'](_intensity))  # type: ignore
        artists_to_remove.append(new_artist)
        return [new_artist]

    ani = FuncAnimation(fig, animate,
                        frames=len(report.solutions),
                        blit=True,  # Supposedly "optimises drawing"
                        repeat=False,
                        interval=400,
                        cache_frame_data=False)

    # See here https://matplotlib.org/stable/api/animation_api.html
    #   for available writers. PillowWriter is chosen here arbitrarily.
    ani.save(file_name,
             dpi=300,
             writer=PillowWriter(fps=25)) # type: ignore[reportArgumentType]

animate_and_save()
```
